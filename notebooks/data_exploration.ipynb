```python
results_df = results[
    ["raceId", "driverId", "positionNumber"]
].rename(columns={"positionNumber": "finish_position"})
```

```python
qualifying_df = qualifying[
    ["raceId", "driverId", "positionNumber"]
].rename(columns={"positionNumber": "qualifying_position"})
```

```python
grid_df = grid[
    ["raceId", "driverId", "positionNumber"]
].rename(columns={"positionNumber": "grid_position"})
```

```python
recent_results= results_df[
    results_df["raceId"].isin(races_2022_2025["id"])
]
recent_qualifying = qualifying_df[
    qualifying_df["raceId"].isin(races_2022_2025["id"])]
print(
    "Results:",
    recent_results.duplicated(["raceId", "driverId"]).sum())
print(
    "Qualifying:",
    recent_qualifying.duplicated(["raceId", "driverId"]).sum())
```

```python
recent_qualifying= recent_qualifying.drop_duplicates(
    subset=["raceId", "driverId"]
)
recent_qualifying.duplicated(
    ["raceId", "driverId"]
).sum()
```

```python
race_data= recent_results.merge(
    recent_qualifying,
    on=["raceId","driverId"],
    how="left",
    validate="one_to_one"

)
```

```python
race_data
```

```python
print(race_data.shape)
display(race_data.head())
print(race_data.isnull().sum())
```

```python
print("Original rows:", len(recent_results))
print("Merged rows:", len(race_data))

print(
    "Duplicates:",
    race_data.duplicated(["raceId", "driverId"]).sum()
)
```
